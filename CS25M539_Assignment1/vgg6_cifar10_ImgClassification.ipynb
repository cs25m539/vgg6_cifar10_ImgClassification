```python
from pathlib import Path
import shutil

backup_root = Path("/content/drive/MyDrive/CS25M539_Assignment1_backup")

try:
    from google.colab import drive
    drive.mount("/content/drive")

    drive_root = Path("/content/drive/MyDrive")
    if drive_root.exists():
        backup_root.parent.mkdir(parents=True, exist_ok=True)
        if backup_root.exists():
            shutil.rmtree(backup_root)
        shutil.copytree(project_dir, backup_root)
        print("Backup created at:", backup_root)
    else:
        print("Google Drive is not mounted; local project remains the primary copy.")
except Exception as exc:
    print("Drive backup skipped:", exc)
    print("All primary files remain in:", project_dir)
```
